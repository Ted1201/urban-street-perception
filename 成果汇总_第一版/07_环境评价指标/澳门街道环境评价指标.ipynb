# 街道环境质量代理评价

研究目标：基于道路属性与街景视觉特征，评价可见绿化、可见机动车和标注限速所反映的环境倾向。联合代理目标 environment_index 与视觉补充目标 visual_subindex 分别统计。研究范围、公式和限制见 outputs/environment_indicators_v1/环境质量代理目标定义.md。

```python
from pathlib import Path
import subprocess,sys,json
import pandas as pd
from IPython.display import display,Image
ROOT=next(p for p in [Path.cwd(),*Path.cwd().parents] if (p/'scripts/build_environment_indicators.py').exists())
subprocess.run([sys.executable,str(ROOT/'scripts/build_environment_indicators.py')],cwd=ROOT,check=True)
OUT=ROOT/'outputs/environment_indicators_v1'
d=pd.read_csv(OUT/'街道环境评价指标全表.csv',encoding='utf-8-sig')
display(pd.Series(json.loads((OUT/'结果摘要.json').read_text(encoding='utf-8'))).to_frame('结果'))
display(d[['road_id','n_green_usable','n_vehicle_usable','green_score','vehicle_score','speed_score','visual_subindex','environment_index','environment_group']].head(15))
display(Image(filename=str(OUT/'环境评价指标分布.png')))
```
